```python
eth_url = "https://drive.google.com/uc?id=1ZMuVs5IW8FuJHWipbF0vWm1L-Xr-F-tg"  
eth = pd.read_csv(eth_url)
print(eth.head())

```

```python
sol_url = "https://drive.google.com/uc?id=1jOH7TcSpB6MKiT9W4VpHCSzZzXyNMlNq"  
sol = pd.read_csv(sol_url)
print(sol.head())
```

```python
xrp_url = "https://drive.google.com/uc?id=1CS3d-1syse7EjpF5n5rjeW1yMJFsYLJx"
xrp = pd.read_csv(xrp_url)
print(xrp.head())
```

```python
btc = btc.iloc[2:]
bnb = bnb.iloc[2:]
eth = eth.iloc[2:]
sol = sol.iloc[2:] 
xrp = xrp.iloc[2:]
```

```python
btc.columns = ["Date","Close","High","Low","Open","Volume"]
bnb.columns = ["Date","Close","High","Low","Open","Volume"]
eth.columns = ["Date","Close","High","Low","Open","Volume"]
sol.columns = ["Date","Close","High","Low","Open","Volume"]
xrp.columns = ["Date","Close","High","Low","Open","Volume"]
```

```python
btc["Date"] = pd.to_datetime(btc["Date"])
bnb["Date"] = pd.to_datetime(bnb["Date"])
eth["Date"] = pd.to_datetime(eth["Date"])
sol["Date"] = pd.to_datetime(sol["Date"])
xrp["Date"] = pd.to_datetime(xrp["Date"])
```

```python
btc = btc.astype({
    "Close": float,
    "High": float,
    "Low": float,
    "Open": float,
    "Volume": float
})
bnb = bnb.astype({
    "Close": float,
    "High": float,
    "Low": float,
    "Open": float,
    "Volume": float
})
eth = eth.astype({
    "Close": float,
    "High": float,
    "Low": float,
    "Open": float,
    "Volume": float
})
sol = sol.astype({
    "Close": float,
    "High": float,
    "Low": float,
    "Open": float,
    "Volume": float
})
xrp = xrp.astype({
    "Close": float,
    "High": float,
    "Low": float,
    "Open": float,
    "Volume": float
})
```

```python
btc = btc.sort_values("Date")
bnb = bnb.sort_values("Date")
eth = eth.sort_values("Date")
sol = sol.sort_values("Date")
xrp = xrp.sort_values("Date")
```

```python
btc.fillna(method="ffill", inplace=True)
bnb.fillna(method="ffill", inplace=True)
eth.fillna(method="ffill", inplace=True)
sol.fillna(method="ffill", inplace=True)
xrp.fillna(method="ffill", inplace=True)
```
